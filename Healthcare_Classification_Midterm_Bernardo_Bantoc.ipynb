```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_classification

# Generate 1,000 records with 5 features and 1 target
X, y = make_classification(
    n_samples=1000,
    n_features=5,
    n_informative=4,
    n_redundant=0,
    n_classes=2,
    weights=[0.68, 0.32],
    random_state=42
)

# Convert the generated features into understandable ranges
df = pd.DataFrame({
    "age": np.round(18 + (X[:, 0] - X[:, 0].min()) /
                    (X[:, 0].max() - X[:, 0].min()) * 72).astype(int),
    "blood_pressure": np.round(90 + (X[:, 1] - X[:, 1].min()) /
                    (X[:, 1].max() - X[:, 1].min()) * 90).astype(int),
    "glucose_level": np.round(70 + (X[:, 2] - X[:, 2].min()) /
                    (X[:, 2].max() - X[:, 2].min()) * 230).astype(int),
    "bmi": np.round(16 + (X[:, 3] - X[:, 3].min()) /
                    (X[:, 3].max() - X[:, 3].min()) * 24, 1),
    "hospital_visits": np.round((X[:, 4] - X[:, 4].min()) /
                    (X[:, 4].max() - X[:, 4].min()) * 10).astype(int),
    "high_risk": y.astype(int)
})

# Save the dataset as a CSV file
df.to_csv("healthcare_risk_classification.csv", index=False)

print("Dataset successfully created!")
print("Rows and columns:", df.shape)
print("Number of features:", 5)
print("Response variable: high_risk")
df.head()
```
